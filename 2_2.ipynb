## Проверяемая гипотеза

В выбранные двое суток концентрация PM2.5 различается между станциями и периодами суток,
а при слабом ветре её средний уровень выше, чем при более сильном ветре.

Гипотеза разделена на три части, чтобы вывод по одной части не подменял выводы по остальным.
Слабым считается ветер менее 2 м/с; это рабочая граница группировки, а не норматив качества воздуха.
Для территориального сравнения используются пары измерений за одинаковые часы.


```python
checks = pd.DataFrame([
    ('Территория', 'Среднее PM2.5 на двух станциях различается',
     'Средние, медианы и почасовые парные разности', '48 пар часов'),
    ('Время', 'Среднее PM2.5 различается между периодами суток',
     'Среднее по периоду суток отдельно на каждой станции и дате', 'Две даты; повторяемость ограничена'),
    ('Ветер', 'Среднее PM2.5 при ветре менее 2 м/с выше',
     'Средние по группам и описательная корреляция Спирмена по станциям', 'Связь не доказывает причину'),
], columns=['Часть гипотезы', 'Ожидание', 'Способ проверки', 'Ограничение'])
display(checks)
```

## Правила интерпретации

Почасовые значения зависимы во времени, а у двух станций общий городской погодный фон.
Поэтому 96 строк не считаются 96 независимыми экспериментами. Различия будут описаны численно,
без обещания статистической значимости и без причинного вывода о транспорте или заводах.

Один город, две станции и двое суток не позволяют проверить сезонность или устойчивость найденной связи.
Годовые показатели стран из XML и год публикации статьи не заменяют даты почасовых измерений.
Превышения ПДК не рассчитываются: для них сначала нужно выбрать норматив и соответствующий период усреднения.


```python
def daypart(hour):
    return ['Ночь', 'Утро', 'День', 'Вечер'][hour // 6]

calendar = pd.DataFrame({'hour': range(24)})
calendar['daypart'] = calendar.hour.map(daypart)
display(calendar.groupby('daypart', sort=False).agg(
    первый_час=('hour', 'min'), последний_час=('hour', 'max'), число_часов=('hour', 'size')))
assert calendar.daypart.nunique() == 4
```
